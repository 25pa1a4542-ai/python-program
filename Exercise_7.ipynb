```python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}
#Union
print("Union:", A | B)
#Intersection
print("Intersection:", A & B)
#Difference
print("Difference:", A - B)
#Symmetric difference
print("Symmetric difference:", A ^ B)
```

```python
A = {1, 2, 3, 4}
B = {3, 4, 5}
#1.Add
A.add(6)
print("add:", A)
#2.update
A.update([7,8])
print(" update:", A)
#3.remove
A.remove(8)
print(" remove:", A)
#4.discard
A.discard(7)
print("discard:", A)
#5.pop
A.pop()
print("pop:", A)
#6.clear
A.clear()
print("clear:", A)
#7.union
print("union:", A.union(B))
#8.intersection
print("intersection:", A.intersection(B))
#9.difference
print("difference:", A.difference(B))
#10.symmetric difference
print("symmetric difference:", A.symmetric_difference(B))
```

```python
#Write a python program that prints missing numbers between 1 to n
n = int(input("Enter n: "))
L = [int(i) for i in input().split()]
L1=set(L)
print(L1)
elements = set(range(1, n+1))
print(elements)
print(elements-L1)
```
