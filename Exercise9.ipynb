```python
#program on "w" mode
file = open("new.txt", "w")
data = "python program"
file.write(data)
print("file created")
file.close()
```

```python
#program on "r" mode
with open("new.txt","r") as file:
  data = file.read()
  print(data)
```

```python
#prgram on "a" mode
with open("new.txt","a") as file:
  file.writelines(["\napple", "\nname", "\nend"])
```

```python
#program on "r" with readline
with open("new.txt","r") as file:
  print(file.readline())
```

```python
with open ("new.txt","r") as file:
  print(file.readline())
  print(file.tell())
  file.seek(1)
  print(file.tell())
  print(file.readline())
```

```python
#copying content of one file to other
with open("new.txt","r") as file:
  data = file.read()
  print("data")
  with open("new1.txt","w") as file1:
    file1.write(data)
    print("data copied")
```

```python
#program that prints each line in a file in reverse
with open("new.txt","r") as file:
  data = file.readlines()
  for i in data:
    print(i[::-1])
```

```python
#write a program to sort words in a file
#file should have only lower-case words,
#must be lowered
with open("new.txt", "r") as file:
  data = file.read().split()
  data_lower = [i.lower() for i in data]
  sort_data=sorted(data_lower)
with open("new1.txt", "w") as file1:
  for i in sort_data:
    file1.write(i+"\n")
print("done")
```

```python
#python program to compute the number of characters, words and lines in a file

f = open("sample.txt", "w")
f.write("Hello Python\nWelcome to programming")
f.close()

f = open("sample.txt", "r")
data = f.read()

print("Characters:", len(data))
print("Words:", len(data.split()))
print("Lines:", len(data.splitlines()))

f.close()
```
