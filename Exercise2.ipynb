```python
# 2. program to find the year is leap year or not
year=int(input("enter a year"))
if(year%4==0 and year%100!=0 or year%400==0):
  print("year is leap year",year)
else:
  print("year is not a leap year",year)
```

```python
# 3. explain code on match-case
day=int(input())
match day:
  case 1:
    print("sunday")
  case 2:
    print("monday")
  case 3:
    print("tuesday")
  case 4:
    print("wednesday")
  case 5:
    print("thursday")
  case 6:
    print("friday")
  case 7:
    print("saturday")
  case _:
    print("invalid")
```

```python
# 12. program that finds sum of digits of a given number
n=int(input("enter a number"))
sum=0
while(n>0):
  r=n%10
  sum=sum+r
  n=n//10
print("sum of the given number is",sum)
```

```python
# 13. program that checks give number is prime number or not
n=int(input("enter a number"))
for i  in range(2,n):
   if(n%i==0):
    print("not a prime number")
    break
else:
    print("prime number")
```
