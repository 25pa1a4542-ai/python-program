```python
# Write a program that counts number of vowles in a string
s = input("enter a string:")
vowels = "aeiou"
vowel_count = 0
for ch in s:
  if ch.lower()in vowels :
    vowel_count += 1
print("number of vowels:",vowel_count)

```

```python
#Program that prints reverse of the number using function
def reverse(n):
  r = 0
  while n > 0:
    d = n % 10
    r = r * 10 +d
    n = n // 10
  return r
n = int(input("enter a number:"))
print("reverse =",reverse(n))
```
