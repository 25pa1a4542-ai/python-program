```python

# Built-in Functions used on Dictionaries

d = {"a": 10, "b": 20, "c": 30}

print("Dictionary Operations")

# 1. len()
print("Length:", len(d))

# 2. sorted()
print("Sorted keys:", sorted(d))

# 3. max()
print("Maximum key:", max(d))

# 4. min()
print("Minimum key:", min(d))

# 5. list()
print("List of keys:", list(d))

# 6. dict()
x = dict(a=10, b=20)
print("New dictionary:", x)

# 7. any()
print("Any key exists:", any(d))

# 8. all()
print("All keys exist:", all(d))

# 9. str()
print("String representation:", str(d))

# 10. type()
print("Type:", type(d))
```

```python
#Dictionary Operations
d = {"name": "pravallika", "age": 18, "branch":"AI&DS"}
#1.Access
print("Name:", d["name"])
#2.Add
d["city"] = "kakinada"
print("after adding:",d)
#3.Update
d["age"] = 20
print("After updating:",d)
#4.Membership
print("name" in d)
#5.Length
print("Length:", len(d))
#6.Delete
del d["branch"]
print("After deleting:",d)
#7.Iteration
print("Dictionary elements:")
for key in d:
  print(key, ":", d[key])
```

```python
# Write a program to check if a given key exists in a dictionary or not
d={"name": "sita", "age": 19, "city": "kakinada"}
key = input("Enter key:")
if key in d:
  print("Yes")
else:
  print("No")
```
